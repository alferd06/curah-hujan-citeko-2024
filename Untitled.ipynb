```python
# Extract Data
import pandas as pd
import glob

# Mengumpulukan nama path tiap data -- list of string
all_name_files = glob.glob('data/**/*.csv')

# baca csv dan menampungnya ke list -- list of dataframe
cd = []

for file in all_name_files:
    df_temp = pd.read_csv(file, skiprows = 7)
    df_temp = df_temp.iloc[:-7]
    cd.append(df_temp)

# Penggabungan semua tabel dan Pembersihan kolom yang tidak dibutuhkan
df = pd.concat(cd, ignore_index = True).drop(columns=['Unnamed: 2'])
print("tabel sudah dimuat dan dibesrsihkan")
```

```python
# Cek Kelengkapan Data
print(f'tabel memiliki {df.shape[0]} baris dan {df.shape[1]} kolom')
# print(df.dtypes)

# tanggal masih str, ubah ke date
df['TANGGAL'] = pd.to_datetime(df['TANGGAL'], format='%d-%m-%Y')
# sort tanggal, hapus dan reset index
df = df.sort_values('TANGGAL').reset_index(drop=True)

# 8888, 9999 -> NaN
df['RR'] = df['RR'].replace([9999, 8888], pd.NA)
# setelah menjadi NaN, tipe data kolom mix
# konversi ke numerik
df['RR'] = pd.to_numeric(df['RR'], errors='coerce')
df['RR'].dtype
# NA -> interpolasi
df['RR'] = df['RR'].interpolate(method='linear')

# df.head(50)
```

```python
import matplotlib.pyplot as plt

# Visualisai Data
plt.figure(figsize=(12, 5))
plt.plot(df['TANGGAL'], df['RR'], color='steelblue')
plt.title('Curah Hujan Citeko 2024 - 2026')
plt.xlabel('Periode')
plt.ylabel('Curah Hujan (mm)')
plt.show()
```

```python
# Agregasi per bulan
df['BULAN'] = df['TANGGAL'].dt.to_period('M')
df_bulanan = df.groupby('BULAN')['RR'].sum().reset_index()
df_bulanan['BULAN'] = df_bulanan['BULAN'].astype(str)

# Visualisasi bulanan
plt.figure(figsize=(14, 5))
plt.bar(df_bulanan['BULAN'], df_bulanan['RR'], color='steelblue')
plt.title('Total Curah Hujan Bulanan Citeko 2024 - 2026')
plt.xlabel('Bulan')
plt.ylabel('Total Curah Hujan (mm)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Tentukan threshold ekstrem -- atas 95% dari semua data
threshold = df['RR'].quantile(0.95)
print(f'Threshold curah hujan ekstrem: {threshold:.2f} mm')

# Filter hari ekstrem
df_ekstrem = df[df['RR'] >= threshold]
print(f'Jumlah hari ekstrem: {len(df_ekstrem)}')
print(df_ekstrem[['TANGGAL', 'RR']].sort_values('RR', ascending=False))

# Visualisasi
plt.figure(figsize=(14, 5))
plt.plot(df['TANGGAL'], df['RR'], color='steelblue', alpha=0.5, label='Harian')
plt.scatter(df_ekstrem['TANGGAL'], df_ekstrem['RR'], 
            color='red', zorder=5, label=f'Ekstrem (≥{threshold:.1f}mm)')
plt.axhline(y=threshold, color='red', linestyle='--', alpha=0.5)
plt.title('Curah Hujan Harian & Hari Ekstrem Citeko 2024-2026')
plt.xlabel('Periode')
plt.ylabel('Curah Hujan (mm)')
plt.legend()
plt.tight_layout()
plt.show()
```
