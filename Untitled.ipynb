```python
# Extract Data
import pandas as pd
import glob

# Mengumpulukan nama path tiap data -- list of string
all_name_files = glob.glob('data/**/*.csv')

# baca csv dan menampungnya ke list -- list of dataframe
cd = []

for file in all_name_files:
    df_temp = pd.read_csv(file, skiprows = 7)
    df_temp = df_temp.iloc[:-7]
    cd.append(df_temp)

# Penggabungan semua tabel dan Pembersihan kolom yang tidak dibutuhkan
df = pd.concat(cd, ignore_index = True).drop(columns=['Unnamed: 2'])
print("tabel sudah dimuat dan dibesrsihkan")
```

```python
# Cek Kelengkapan Data
print(f'tabel memiliki {df.shape[0]} baris dan {df.shape[1]} kolom')
# print(df.dtypes)

# tanggal masih str, ubah ke date
df['TANGGAL'] = pd.to_datetime(df['TANGGAL'], format='%d-%m-%Y')
# sort tanggal, hapus dan reset index
df = df.sort_values('TANGGAL').reset_index(drop=True)

# 8888, 9999 -> NaN
df['RR'] = df['RR'].replace([9999, 8888], pd.NA)
# setelah menjadi NaN, tipe data kolom mix
# konversi ke numerik
df['RR'] = pd.to_numeric(df['RR'], errors='coerce')
df['RR'].dtype
# NA -> interpolasi
df['RR'] = df['RR'].interpolate(method='linear')

# df.head(50)
```

```python
import matplotlib.pyplot as plt

# Visualisai Data
plt.figure(figsize=(12, 5))
plt.plot(df['TANGGAL'], df['RR'], color='steelblue')
plt.title('Curah Hujan Citeko 2024 - 2026')
plt.xlabel('Periode')
plt.ylabel('Curah Hujan (mm)')
plt.show()
```
