```python
import pandas as pd
import matplotlib.pyplot as plt
print("Siap!")
```

```python
df = pd.read_csv("data/laporan_iklim_harian.csv", skiprows=7)
df.head(10)
df.shape
df = df.iloc[:-7]
df.tail(10)
```

```python
df = df.drop(columns = ['Unnamed: 2'])
```

```python
df['RR'] = pd.to_numeric(df['RR'], errors='coerce')
df['RR'] = df['RR'].replace([8888, 9999], pd.NA)
print(df['RR'].isna().sum())
df.head(10)
print(df.isna().sum())
```

```python
df['RR'] = df['RR'].fillna(df['RR'].mean())
df['RR'].isna().sum()
```

```python
plt.figure(figsize=(12, 5))
plt.plot(df['TANGGAL'], df['RR'], marker = 'o', color='steelblue')
plt.title('Curah Hujan Harian Juni - Juli 2024 (Stasiun Citeko)')
plt.xlabel('Tanggal')
plt.ylabel('Curah Hujan (mm)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df.to_csv('data/juni_2024_bersih.csv')
print('Tersimpan')
```

```python
import seaborn as sns

plt.figure(figsize=(12, 5))
sns.lineplot(data=df, x='TANGGAL', y='RR', marker='o')
plt.title('Curah Hujan Harian')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Tanpa figure baru di tengah
plt.plot(df['TANGGAL'], df['RR'], color='steelblue', label='data 1')
plt.plot(df['TANGGAL'], df['RR'] * 2, color='red', label='data 2')
plt.legend()
plt.show()
```
